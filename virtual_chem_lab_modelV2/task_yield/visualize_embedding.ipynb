```python
query_embedding_umap = np.load("/data1/ReactionT5_neword/task_yield/output_test1_test/query_embedding_umap.npy")

# calculate the molecular weight
from rdkit import Chem
from rdkit.Chem import Descriptors
# silence warning
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.warning')
from transformers import AutoTokenizer 
tokenizer = AutoTokenizer.from_pretrained("sagawa/ReactionT5v2-yield")

for df in [df_target, df_query]:
    df["mol"] = df["input"].apply(Chem.MolFromSmiles)
    df["mw"] = df["mol"].apply(Descriptors.MolWt)
    df["logp"] = df["mol"].apply(Descriptors.MolLogP)
    df["hba"] = df["mol"].apply(Descriptors.NumHAcceptors)
    df["hbd"] = df["mol"].apply(Descriptors.NumHDonors)
    df["tpsa"] = df["mol"].apply(Descriptors.TPSA)
    df["rotb"] = df["mol"].apply(Descriptors.NumRotatableBonds)
    df["atoms"] = df["mol"].apply(Descriptors.HeavyAtomCount)
    df["rings"] = df["mol"].apply(Descriptors.RingCount)
    df["qed"] = df["mol"].apply(Descriptors.qed)
    df["input_len"] = df["input"].apply(lambda x: len(tokenizer(x)["input_ids"]))
```

```python
# for each feature, clip with 5% and 95% percentile
features = ["mw", "logp", "hba", "hbd", "tpsa", "rotb", "atoms", "rings", "qed", "input_len"]
for feature in features:
    lower = df_target[feature].quantile(0.05)
    upper = df_target[feature].quantile(0.95)
    df_target[feature] = df_target[feature].clip(lower, upper)
    lower = df_query[feature].quantile(0.05)
    upper = df_query[feature].quantile(0.95)
    df_query[feature] = df_query[feature].clip(lower, upper)
```

```python
# from rdkit import DataStructs
# similarity_score1 = [[0 for __ in range(len(target_fp1))] for _ in range(len(query_fp1))]
# for i, query in enumerate(query_fp1):
#     for j, target in enumerate(target_fp1):
#         similarity_score1[i][j] = DataStructs.TanimotoSimilarity(query, target)
# similarity_score1 = np.array(similarity_score1)
# np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test1_test/similarity_score_morgan_tanimoto.npy", similarity_score1)
# del similarity_score1
# import gc
# gc.collect()

# similarity_score2 = [[0 for __ in range(len(target_fp1))] for _ in range(len(query_fp2))]
# for i, query in enumerate(query_fp2):
#     for j, target in enumerate(target_fp1):
#         similarity_score2[i][j] = DataStructs.TanimotoSimilarity(query, target)

# similarity_score2 = np.array(similarity_score2)
# np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test2_test/similarity_score_morgan_tanimoto.npy", similarity_score2)
# del similarity_score2
# gc.collect()

# similarity_score3 = [[0 for __ in range(len(target_fp1))] for _ in range(len(query_fp3))]
# for i, query in enumerate(query_fp3):
#     for j, target in enumerate(target_fp1):
#         similarity_score3[i][j] = DataStructs.TanimotoSimilarity(query, target)

# similarity_score3 = np.array(similarity_score3)
# np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test3_test/similarity_score_morgan_tanimoto.npy", similarity_score3)
# del similarity_score3
# gc.collect()

# similarity_score4 = [[0 for __ in range(len(target_fp1))] for _ in range(len(query_fp4))]
# for i, query in enumerate(query_fp4):
#     for j, target in enumerate(target_fp1):
#         similarity_score4[i][j] = DataStructs.TanimotoSimilarity(query, target)

# similarity_score4 = np.array(similarity_score4)
# np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test4_test/similarity_score_morgan_tanimoto.npy", similarity_score4)
# del similarity_score4
# gc.collect()

import numpy as np
from rdkit import DataStructs
from multiprocessing import Pool, cpu_count

def compute_tanimoto_chunk(args):
    """
    チャンクの開始インデックスと終了インデックス、query_fps、target_fps が引数。
    Tanimoto 類似度を計算した 2次元配列を返す。
    """
    start_idx, end_idx, query_fps, target_fps = args
    chunk_result = []
    for i in range(start_idx, end_idx):
        qfp = query_fps[i]
        # target_fps 全体と比較した 1 行分 (list) を作成
        row = [DataStructs.TanimotoSimilarity(qfp, tfp) for tfp in target_fps]
        chunk_result.append(row)
    # 開始インデックスと計算結果をタプルで返す
    return (start_idx, np.array(chunk_result))

def parallel_tanimoto_similarity(query_fps, target_fps, n_processes=None):
    """
    query_fps と target_fps の Tanimoto 類似度スコア行列を並列計算して numpy array で返す。
    n_processes を指定しない場合、CPU コア数を自動で取得。
    """
    if n_processes is None:
        n_processes = cpu_count()  # またはお好みのコア数に固定
    
    n_queries = len(query_fps)
    similarity_score = np.zeros((n_queries, len(target_fps)), dtype=np.float32)

    # チャンク分割数。大きすぎると並列効率が落ち、小さすぎるとオーバーヘッドが増える。
    # 例えば n_processes の数倍〜数十倍くらいがほどよい。
    # ここでは一例として n_processes の10倍のチャンクにしてみる
    chunks = n_processes * 40

    chunk_size = (n_queries + chunks - 1) // chunks  # 切り上げ
    tasks = []
    for i in range(chunks):
        start_idx = i * chunk_size
        end_idx = min((i+1) * chunk_size, n_queries)
        if start_idx >= n_queries:
            break
        tasks.append((start_idx, end_idx, query_fps, target_fps))
    
    # 並列実行
    with Pool(n_processes) as p:
        results = p.map(compute_tanimoto_chunk, tasks)
    
    # 結果を結合：各チャンクごとの計算結果を適切な位置に配置
    for start_idx, chunk_res in results:
        similarity_score[start_idx : start_idx + chunk_res.shape[0], :] = chunk_res
    
    return similarity_score

# --- 実際の呼び出し例 ---
# query_fp1 のみ例示。他の query_fp2, query_fp3, query_fp4 も同様に呼び出し可能
similarity_score1 = parallel_tanimoto_similarity(query_fp1, target_fp1, n_processes=8)
np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test1_test/similarity_score_morgan_tanimoto.npy",
        similarity_score1)

# 必要に応じてメモリ開放
del similarity_score1
import gc
gc.collect()

# 以下、query_fp2, query_fp3, query_fp4 についても同様の処理を実施
similarity_score2 = parallel_tanimoto_similarity(query_fp2, target_fp1, n_processes=8)
np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test2_test/similarity_score_morgan_tanimoto.npy",
        similarity_score2)
del similarity_score2
gc.collect()

similarity_score3 = parallel_tanimoto_similarity(query_fp3, target_fp1, n_processes=8)
np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test3_test/similarity_score_morgan_tanimoto.npy",
        similarity_score3)
del similarity_score3
gc.collect()

similarity_score4 = parallel_tanimoto_similarity(query_fp4, target_fp1, n_processes=8)
np.save("/mnt/d/ReactionT5_neword_241116/task_yield/output_test4_test/similarity_score_morgan_tanimoto.npy",
        similarity_score4)
del similarity_score4
gc.collect()
```
